```python
# ============================================================
# EXPERIMENT 2: DATA PREPROCESSING
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

df = pd.read_csv("clv_data.csv")

print("========== ORIGINAL DATASET ==========")
print("Number of rows    :", df.shape[0])
print("Number of columns :", df.shape[1])

print("\nFirst 5 rows:")
display(df.head())


# ------------------------------------------------------------
# 2. ANALYZE DATA TYPES
# ------------------------------------------------------------

print("\n========== DATA TYPES ==========")
print(df.dtypes)


print("\nNumerical columns:")
print(df.select_dtypes(include=np.number).columns.tolist())

print("\nText/Categorical columns:")
print(df.select_dtypes(include="object").columns.tolist())


# ------------------------------------------------------------
# 3. CHECK DUPLICATE VALUES
# ------------------------------------------------------------

print("\n========== DUPLICATE ANALYSIS ==========")

duplicate_count = df.duplicated().sum()

print("Duplicate rows before removal:", duplicate_count)

# Remove duplicate rows
df = df.drop_duplicates()

print("Duplicate rows after removal:", df.duplicated().sum())


# ------------------------------------------------------------
# 4. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\n========== MISSING VALUE ANALYSIS ==========")

print("Missing values BEFORE handling:")
print(df.isnull().sum())


# ------------------------------------------------------------
# 5. DISPLAY MISSING VALUE PERCENTAGE
# ------------------------------------------------------------

print("\nMissing value percentage:")
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)


# ------------------------------------------------------------
# 6. SHOW DIFFERENT WAYS OF HANDLING MISSING VALUES
# ------------------------------------------------------------

print("\n========== MISSING VALUE HANDLING METHODS ==========")

# A. Removing rows containing missing values
removed_rows = df.dropna()

print("\nA. Removing rows with missing values")
print("Rows before:", len(df))
print("Rows after :", len(removed_rows))
print("Rows removed:", len(df) - len(removed_rows))


# B. Fill missing numerical values with 0
zero_filled = df.copy()

numeric_columns = zero_filled.select_dtypes(
    include=np.number
).columns

zero_filled[numeric_columns] = zero_filled[numeric_columns].fillna(0)

print("\nB. Filling numerical missing values with 0")
print(zero_filled.isnull().sum())


# ------------------------------------------------------------
# 7. MEAN, MEDIAN AND MODE
# ------------------------------------------------------------

print("\n========== MEAN, MEDIAN AND MODE ==========")

for column in df.select_dtypes(include=np.number).columns:

    if df[column].isnull().sum() > 0:

        print("\nColumn:", column)

        print("Mean   :", df[column].mean())
        print("Median :", df[column].median())
        print("Mode   :", df[column].mode()[0])


# ------------------------------------------------------------
# 8. SHOW BEFORE VALUE
# ------------------------------------------------------------

print("\n========== BEFORE MISSING VALUE HANDLING ==========")

print(df[["age", "days_on_platform"]].head(10))


# ------------------------------------------------------------
# 9. CREATE REFINED DATASET
# ------------------------------------------------------------

refined_df = df.copy()


# Fill numerical missing values using MEDIAN
# Median is less affected by extreme values.

for column in refined_df.select_dtypes(include=np.number).columns:

    if refined_df[column].isnull().sum() > 0:
        refined_df[column] = refined_df[column].fillna(
            refined_df[column].median()
        )


# Fill categorical missing values using MODE

for column in refined_df.select_dtypes(include="object").columns:

    if refined_df[column].isnull().sum() > 0:
        refined_df[column] = refined_df[column].fillna(
            refined_df[column].mode()[0]
        )


# ------------------------------------------------------------
# 10. SHOW AFTER VALUE
# ------------------------------------------------------------

print("\n========== AFTER MISSING VALUE HANDLING ==========")

print(refined_df[["age", "days_on_platform"]].head(10))


# ------------------------------------------------------------
# 11. CHECK MISSING VALUES AFTER HANDLING
# ------------------------------------------------------------

print("\nMissing values AFTER handling:")
print(refined_df.isnull().sum())


# ------------------------------------------------------------
# 12. ONE-HOT ENCODING
# ------------------------------------------------------------

print("\n========== ONE-HOT ENCODING ==========")

print("Categorical columns before encoding:")
print(refined_df.select_dtypes(include="object").columns.tolist())

# Convert text columns into numerical columns
refined_df = pd.get_dummies(
    refined_df,
    columns=refined_df.select_dtypes(include="object").columns,
    drop_first=True
)

# Convert True/False into 1/0
refined_df = refined_df.astype(int)

print("\nDataset after One-Hot Encoding:")
display(refined_df.head())

print("\nNew data types:")
print(refined_df.dtypes)


# ------------------------------------------------------------
# 13. OUTLIER ANALYSIS USING IQR
# ------------------------------------------------------------

print("\n========== OUTLIER ANALYSIS ==========")

numeric_columns = refined_df.select_dtypes(
    include=np.number
).columns

outlier_summary = []

for column in numeric_columns:

    Q1 = refined_df[column].quantile(0.25)
    Q3 = refined_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = refined_df[
        (refined_df[column] < lower_limit) |
        (refined_df[column] > upper_limit)
    ]

    outlier_summary.append([
        column,
        Q1,
        Q3,
        IQR,
        len(outliers)
    ])

outlier_table = pd.DataFrame(
    outlier_summary,
    columns=[
        "Feature",
        "Q1",
        "Q3",
        "IQR",
        "Number of Outliers"
    ]
)

display(outlier_table)


# ------------------------------------------------------------
# 14. BOXPLOT BEFORE OUTLIER REMOVAL
# ------------------------------------------------------------

print("\n========== BOXPLOT BEFORE OUTLIER REMOVAL ==========")

# Original important numerical columns
plot_columns = [
    "age",
    "income",
    "days_on_platform",
    "purchases"
]

for column in plot_columns:

    if column in refined_df.columns:

        plt.figure(figsize=(7, 4))

        sns.boxplot(y=refined_df[column])

        plt.title("Boxplot Before Outlier Removal - " + column)
        plt.ylabel(column)

        plt.show()


# ------------------------------------------------------------
# 15. REMOVE OUTLIERS USING IQR
# ------------------------------------------------------------

print("\n========== REMOVING OUTLIERS ==========")

before_outliers = len(refined_df)

for column in numeric_columns:

    Q1 = refined_df[column].quantile(0.25)
    Q3 = refined_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    refined_df = refined_df[
        (refined_df[column] >= lower_limit) &
        (refined_df[column] <= upper_limit)
    ]

after_outliers = len(refined_df)

print("Rows before outlier removal:", before_outliers)
print("Rows after outlier removal :", after_outliers)
print("Outliers removed            :", before_outliers - after_outliers)


# ------------------------------------------------------------
# 16. BOXPLOT AFTER OUTLIER REMOVAL
# ------------------------------------------------------------

print("\n========== BOXPLOT AFTER OUTLIER REMOVAL ==========")

for column in plot_columns:

    if column in refined_df.columns:

        plt.figure(figsize=(7, 4))

        sns.boxplot(y=refined_df[column])

        plt.title("Boxplot After Outlier Removal - " + column)
        plt.ylabel(column)

        plt.show()


# ------------------------------------------------------------
# 17. FINAL REFINED DATASET
# ------------------------------------------------------------

print("\n========== FINAL REFINED DATASET ==========")

print("Original number of rows :", df.shape[0])
print("Final number of rows    :", refined_df.shape[0])

print("Original number of columns :", df.shape[1])
print("Final number of columns   :", refined_df.shape[1])

print("\nMissing values in refined dataset:")
print(refined_df.isnull().sum().sum())

print("\nDuplicate rows in refined dataset:")
print(refined_df.duplicated().sum())

print("\nFirst 5 rows of refined dataset:")
display(refined_df.head())


# ------------------------------------------------------------
# 18. SAVE REFINED DATASET
# ------------------------------------------------------------

refined_df.to_csv(
    "refined_clv_data.csv",
    index=False
)

print("\nRefined dataset saved as: refined_clv_data.csv")
```

```python
# ============================================================
# EXPERIMENT 3: MOBILE PRICE PREDICTION
# ============================================================

# Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

# ------------------------------------------------------------
# 1. Load the dataset
# ------------------------------------------------------------

df = pd.read_csv("mobile_price_500.csv")

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Display first 5 rows
print("\nFirst 5 rows:")
display(df.head())

# ------------------------------------------------------------
# 2. Convert 5G into numerical values
# ------------------------------------------------------------

df["5G"] = df["5G"].map({"Yes": 1, "No": 0})

# ------------------------------------------------------------
# 3. Convert Brand into numerical values
# ------------------------------------------------------------

df = pd.get_dummies(df, columns=["Brand"], drop_first=True)

# ------------------------------------------------------------
# 4. Separate input features and target
# ------------------------------------------------------------

X = df.drop("Price", axis=1)
y = df["Price"]

print("\nInput Features:")
print(X.columns)

print("\nTarget Feature: Price")

# ------------------------------------------------------------
# 5. Split dataset into training and testing data
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining data:", X_train.shape)
print("Testing data:", X_test.shape)

# ------------------------------------------------------------
# 6. Scale the features
# ------------------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ------------------------------------------------------------
# 7. MODEL 1 - LINEAR REGRESSION
# ------------------------------------------------------------

linear_model = LinearRegression()

linear_model.fit(X_train_scaled, y_train)

linear_prediction = linear_model.predict(X_test_scaled)

print("\nLinear Regression predictions:")
print(linear_prediction[:5])

# ------------------------------------------------------------
# 8. MODEL 2 - RANDOM FOREST REGRESSION
# ------------------------------------------------------------

random_forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

random_forest_model.fit(X_train, y_train)

random_forest_prediction = random_forest_model.predict(X_test)

print("\nRandom Forest predictions:")
print(random_forest_prediction[:5])

# ------------------------------------------------------------
# 9. Predict the price of a new mobile
# ------------------------------------------------------------

new_mobile = X_test.iloc[0:1]

# Linear Regression prediction
new_mobile_scaled = scaler.transform(new_mobile)

linear_price = linear_model.predict(new_mobile_scaled)

# Random Forest prediction
random_forest_price = random_forest_model.predict(new_mobile)

print("\n======================================")
print("PREDICTED PRICE OF NEW MOBILE")
print("======================================")

print("Linear Regression Price: ₹", round(linear_price[0], 2))

print("Random Forest Price: ₹", round(random_forest_price[0], 2))
```

```python
# ============================================================
# EXPERIMENT 4: MODEL ACCURACY AND COMPARISON
# ============================================================

# Import libraries
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ------------------------------------------------------------
# 1. Calculate Linear Regression performance
# ------------------------------------------------------------

linear_mae = mean_absolute_error(
    y_test,
    linear_prediction
)

linear_mse = mean_squared_error(
    y_test,
    linear_prediction
)

linear_rmse = np.sqrt(linear_mse)

linear_r2 = r2_score(
    y_test,
    linear_prediction
)

# ------------------------------------------------------------
# 2. Calculate Random Forest performance
# ------------------------------------------------------------

rf_mae = mean_absolute_error(
    y_test,
    random_forest_prediction
)

rf_mse = mean_squared_error(
    y_test,
    random_forest_prediction
)

rf_rmse = np.sqrt(rf_mse)

rf_r2 = r2_score(
    y_test,
    random_forest_prediction
)

# ------------------------------------------------------------
# 3. Display model performance
# ------------------------------------------------------------

print("==============================================")
print("MODEL PERFORMANCE")
print("==============================================")

print("\nLinear Regression")
print("MAE  :", round(linear_mae, 2))
print("MSE  :", round(linear_mse, 2))
print("RMSE :", round(linear_rmse, 2))
print("R2 Score :", round(linear_r2, 4))

print("\nRandom Forest Regression")
print("MAE  :", round(rf_mae, 2))
print("MSE  :", round(rf_mse, 2))
print("RMSE :", round(rf_rmse, 2))
print("R2 Score :", round(rf_r2, 4))

# ------------------------------------------------------------
# 4. Create comparison table
# ------------------------------------------------------------

results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest"
    ],

    "MAE": [
        linear_mae,
        rf_mae
    ],

    "MSE": [
        linear_mse,
        rf_mse
    ],

    "RMSE": [
        linear_rmse,
        rf_rmse
    ],

    "R2 Score": [
        linear_r2,
        rf_r2
    ]
})

print("\n==============================================")
print("COMPARISON TABLE")
print("==============================================")

display(results)

# ------------------------------------------------------------
# 5. Accuracy percentage using R2 Score
# ------------------------------------------------------------

linear_accuracy = linear_r2 * 100
rf_accuracy = rf_r2 * 100

print("\nPrediction Accuracy:")
print("Linear Regression:", round(linear_accuracy, 2), "%")
print("Random Forest:", round(rf_accuracy, 2), "%")

# ------------------------------------------------------------
# 6. Accuracy comparison graph
# ------------------------------------------------------------

models = [
    "Linear Regression",
    "Random Forest"
]

accuracy = [
    linear_accuracy,
    rf_accuracy
]

plt.figure(figsize=(7, 5))

plt.bar(models, accuracy)

plt.xlabel("Machine Learning Model")
plt.ylabel("Accuracy (%)")
plt.title("Comparison of Prediction Accuracy")

plt.ylim(0, 100)

plt.show()

# ------------------------------------------------------------
# 7. Identify model with higher R2 score
# ------------------------------------------------------------

if linear_r2 > rf_r2:
    best_model = "Linear Regression"
    best_accuracy = linear_accuracy
else:
    best_model = "Random Forest"
    best_accuracy = rf_accuracy

print("\n==============================================")
print("COMPARATIVE ANALYSIS")
print("==============================================")

print("Model with higher R2 Score:", best_model)
print("Accuracy:", round(best_accuracy, 2), "%")
```
